```python
#1. data cleanig
# step 1.1 basic of the project to read the dataset
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
data=pd.read_csv('python dataset.csv')
data.head()
```

```python
# step 1.2 explore the data types using descriptive stat analaysis
print(data.info())
print(data.describe())
print(data.isnull().sum())
print(data.isnull().sum().sum())
```

```python
#step 1.3
#number of bags columns as convert into "int" after null value fill by median
data['Number.of.Bags'] = data['Number.of.Bags'].fillna(data['Number.of.Bags'].median())

# Then convert to Int64 type
data['Number.of.Bags'] = data['Number.of.Bags'].astype('Int64')
data
print(data.isnull().sum())
```

```python
#STEP 1.4
data['Bag.Weight'] = (
    data['Bag.Weight']
    .astype(str)                # make sure all are strings
    .str.extract(r'(\d+\.?\d*)') # extract numeric part (e.g., "69.5" from "69.5 kg")
    .astype(float)              # convert to float
)
data.info()
data.to_csv('coffeeQuality_cleaned.csv', index=False)
```

```python
#step 1.5
# Altitude Range 
data['altitude_range_meters'] = data['altitude_high_meters'] - data['altitude_low_meters']
print("Altitude Range:\n", data[['altitude_low_meters', 'altitude_high_meters', 'altitude_range_meters']].head(), "\n")
data.to_csv('coffeeQuality.csv', index=False)
```

```python
#step 1.6
# Average Altitude
data['altitude_avg_meters'] = (data['altitude_low_meters'] + data['altitude_high_meters']) / 2
print("Average Altitude:\n", data[['altitude_low_meters', 'altitude_high_meters', 'altitude_avg_meters']].head(), "\n")
data.to_csv('coffeeQuality.csv', index=False)
```

```python
#step 1.7
#the float data type missing value as fill (fillna)
fill_float = ['Quakers', 'altitude_low_meters', 
              'altitude_high_meters',
              'altitude_mean_meters']
for j in fill_float:
    data[j] = data[j].fillna(data[j].mean())
data.to_csv('coffeeQuality.csv', index=False)
print(data.isnull().sum())
```

```python
#step 1.8
#fill the columns to using fillna
#fill the blanks columns using fill or bfill or else fillna
fill_obj=['Owner','Country.of.Origin','Farm.Name','Lot.Number','Mill','ICO.Number','Company','Altitude','Region','Producer','Harvest.Year','Owner.1','Variety','Processing.Method','Color']
for i in fill_obj:
    data[i]=data[i].ffill().bfill()
data
data.to_csv('coffeeQuality.csv', index=False)
print(data.isnull().sum())
```

```python
#step 1.9
# remove the columns


cols_to_remove = [
    'Lot.Number', 'Mill', 'ICO.Number', 'Owner.1', 'Expiration',
    'In.Country.Partner', 'Certification.Address', 'Certification.Contact', 'Certification.Body'
]

# Remove "Unnamed"
for col in data.columns:
    if col.startswith('Unnamed'):
        cols_to_remove.append(col)

#remove mentioned columns
for col in cols_to_remove:
    if col in data.columns:
        data.drop(col, axis=1, inplace=True)

# Remove rows with missing values
data.dropna(inplace=True)

# Check final columns
print("Remaining columns:", data.columns.tolist())
data
```

```python
#STEP 1.10
#to find any columns having special char 
import re
import pandas as pd  # Make sure pandas is imported

# Function to detect special characters in cell values
def has_special_char(value):
    if pd.isnull(value):  # Changed from data.isnull(value) to pd.isnull(value)
        return False
    # Only allow letters, numbers, spaces, dot, comma, colon, dash
    return bool(re.search(r"[^\w\s.,:-]", str(value)))

# Check each object column for special characters
columns_with_special_chars = []

for col in data.select_dtypes(include='object').columns:
    if data[col].apply(has_special_char).any():
        columns_with_special_chars.append(col)

# Output the result
if columns_with_special_chars:
    print("Columns where values contain special characters:")
    for col in columns_with_special_chars:
        print(f" - {col}")
else:
    print("No special characters found in any column values.")
```

```python
#STEP 1.11
#to remove that special char from having columns
# Columns to clean
cols = [
    'Owner', 'Country.of.Origin', 'Farm.Name',
    'Company', 'Altitude', 'Region', 'Producer',
    'Harvest.Year','Processing.Method'
]

# Clean special characters
for col in cols:
    data[col] = data[col].astype(str).str.replace(r'[^\w\s.,:-]', '', regex=True)

# Save cleaned file
data.to_csv("cleaned_coffee_data.csv", index=False)

print("Cleaned and saved as 'cleaned_coffee_data.csv'")
data
```

```python
#step 1.12
#removeing outlier
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import random

numeric_cols = data.select_dtypes(include=['float64', 'int64']).columns
# Boxplot before removing outliers 
plt.figure(figsize=(12, 6))
data[numeric_cols].boxplot()
plt.title("Before Outlier Removal")
plt.xticks(rotation=90)
plt.show()
#Remove outliers using IQR
for col in numeric_cols:
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    data = data[(data[col] >= lower_bound) & (data[col] <= upper_bound)]

#Boxplot after removing outliers
plt.figure(figsize=(12, 6))
data[numeric_cols].boxplot()
plt.title("After Outlier Removal")
plt.xticks(rotation=90)
plt.show()

# Print shape after cleaning
print("After removing outliers:", data.shape)

# Save the cleaned data
data.to_csv('coffeeQuality.csv', index=False)
```

```python
#statical test
#3. TEST COMPARE
# 3.1 T-TEST
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

# Significance level
alpha = 0.05

# Drop rows with missing values in both columns
paired_data = data[['Aroma', 'Flavor']]

# Paired T-test (samples are related)
t_stat, p_val = stats.ttest_rel(paired_data['Aroma'], paired_data['Flavor'])
print(f"Paired T-test p-value: {p_val:.4f}")

if p_val < alpha:
    print("Reject H0 → Aroma and Flavor have significantly different mean scores.")
else:
    print("Fail to Reject H0 → No significant difference between Aroma and Flavor.")

plt.figure(figsize=(8, 5))
sns.boxplot(data=paired_data[['Aroma', 'Flavor']])
plt.title("Comparison of Aroma and Flavor Scores")
plt.ylabel("Scores")
plt.show()

aroma_mean = data['Aroma'].mean()
flavor_mean = data['Flavor'].mean()
print(f"Mean Aroma Score: {aroma_mean:.2f}")
print(f"Mean Flavor Score: {flavor_mean:.2f}")
```

```python
# 3.2 ANOVA
groups = [g['Aroma'].dropna() for _, g in data.groupby('Processing.Method')]
f_stat, p_val_anova = stats.f_oneway(*groups)
print(f"ANOVA p-value: {p_val_anova:.4f}")

if p_val_anova < alpha:
    print("Reject H0 → Processing methods have significantly different Aroma scores.")
else:
    print("Fail to Reject H0 → No significant difference among processing methods.")

# Visualization
sns.boxplot(x='Processing.Method', y='Aroma', data=data)
plt.title("ANOVA: Aroma by Processing Method")
plt.xticks(rotation=45)
plt.show()
```

```python
# 3.3 CHI SQURE
# Chi-Square: Species vs Country.of.Origin
import pandas as pd
from scipy.stats import chi2_contingency
import seaborn as sns
import matplotlib.pyplot as plt

table2 = pd.crosstab(data['Species'], data['Country.of.Origin'])
chi2, p, dof, ex = chi2_contingency(table2)
print(f"Species vs Country.of.Origin p-value: {p:.4f}")

if p < 0.05:
    print("Reject H0 → Significant association between Species and Country of Origin.")
else:
    print("Fail to Reject H0 → No significant association between Species and Country of Origin.")

# Visualization
plt.figure(figsize=(10, 5))
sns.heatmap(table2, annot=True, fmt="d", cmap="YlGnBu")
plt.title("Chi-Square: Species vs Country of Origin")
plt.show()
```
